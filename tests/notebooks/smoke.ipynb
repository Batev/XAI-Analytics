```python
import numpy as np
import pandas as pd
from IPython.display import display
from util import commons as c, ui
from util.model import ModelType, ProblemType, Algorithm
from util.split import Split, SplitTypes
X = pd.DataFrame({'a': np.arange(40.), 'b': np.tile([0., 1.], 20)})
y = pd.Series(np.tile([0, 1], 20))
mt = ModelType(ProblemType.CLASSIFICATION)
mt.algorithm = Algorithm.DECISION_TREE
pipeline, train, yt, test, ytest = c.train_model(mt, Split(SplitTypes.NORMAL), X, y, search_options={'param_grid': {}, 'cv': 2, 'n_jobs': 1})
assert pipeline.predict(test).shape == ytest.shape
models, message = c.fill_empty_models(X, y, 1)
display(ui.generate_feature_importance_grid(models))
display(ui.generate_local_interpretation_grid(models))
```
